### Exempel 1 — Din första klass och ditt första objekt

En **klass** är en mall som beskriver vad ett objekt ska ha för egenskaper (attribut).
Ett **objekt** är ett konkret exemplar skapat utifrån klassen.

Här skapar vi klassen `Hund` med attributen `namn` och `ras`, och skapar sedan ett objekt av den.

```python
class Hund:
    def __init__(self, namn, ras):
        self.namn = namn
        self.ras = ras

# Skapa ett objekt av klassen Hund
nero = Hund("Nero", "Labrador")

print(nero.namn)
print(nero.ras)
```

### Exempel 2 — Konstruktorn `__init__` och `self`

`__init__` kallas **konstruktorn**. Den körs automatiskt när ett nytt objekt skapas, och sätter startvärdena på attributen.

`self` syftar på objektet självt — `self.märke` betyder "det här specifika objektets märke".

Här skapar vi klassen `Bil`.

```python
class Bil:
    def __init__(self, marke, farg):
        self.marke = marke
        self.farg = farg

volvo = Bil("Volvo", "röd")
saab = Bil("Saab", "blå")

print(volvo.marke, volvo.farg)
print(saab.marke, saab.farg)
```

### Exempel 3 — Metoder

En **metod** är en funktion som hör till en klass. Den skrivs precis som en vanlig funktion, men ligger inuti klassen och tar alltid `self` som första parameter.

Här lägger vi till metoden `skall()` till klassen `Hund`.

```python
class Hund:
    def __init__(self, namn, ras):
        self.namn = namn
        self.ras = ras

    def skall(self):
        print(self.namn + " skäller: Voff!")

nero = Hund("Nero", "Labrador")
nero.skall()
```

### Exempel 4 — Metoder som returnerar ett värde

Metoder kan, precis som vanliga funktioner, använda `return` för att skicka tillbaka ett värde istället för att bara skriva ut något.

Här beräknar klassen `Rektangel` sin area.

```python
class Rektangel:
    def __init__(self, bredd, hojd):
        self.bredd = bredd
        self.hojd = hojd

    def area(self):
        return self.bredd * self.hojd

r = Rektangel(4, 5)
print("Arean är:", r.area())
```

### Exempel 5 — Flera objekt i en lista

Samma klass kan användas för att skapa hur många objekt som helst. Man kan lagra dem i en lista och gå igenom dem med en loop, precis som med vilken lista som helst.

```python
class Hund:
    def __init__(self, namn, ras):
        self.namn = namn
        self.ras = ras

    def skall(self):
        print(self.namn + " skäller: Voff!")

hundar = [Hund("Nero", "Labrador"), Hund("Buster", "Pudel"), Hund("Bella", "Beagle")]

for h in hundar:
    h.skall()
```

### Exempel 6 — Standardvärden i konstruktorn

Precis som vanliga funktioner kan `__init__` ha **standardvärden** (default) för parametrar. Då blir de valfria att ange när man skapar objektet.

```python
class Pizza:
    def __init__(self, topping="ost"):
        self.topping = topping

p1 = Pizza()             # använder standardvärdet "ost"
p2 = Pizza("skinka")     # anger eget värde

print(p1.topping)
print(p2.topping)
```

### Exempel 7 — Att ändra ett attribut efter att objektet skapats

Metoder kan ändra objektets egna attribut, inte bara läsa dem. Här kan man sätta in och ta ut pengar från ett bankkonto — saldot (`self.saldo`) uppdateras varje gång.

```python
class Bankkonto:
    def __init__(self, saldo=0):
        self.saldo = saldo

    def satt_in(self, belopp):
        self.saldo = self.saldo + belopp

    def ta_ut(self, belopp):
        self.saldo = self.saldo - belopp

konto = Bankkonto(100)
konto.satt_in(50)
konto.ta_ut(30)

print("Saldo:", konto.saldo)
```

### Exempel 8 — `__str__`: snygg utskrift av ett objekt

Om man skriver `print(mitt_objekt)` utan en `__str__`-metod får man ett kryptiskt meddelande. Med `__str__` bestämmer man själv vad som ska visas.

```python
class Bok:
    def __init__(self, titel, forfattare):
        self.titel = titel
        self.forfattare = forfattare

    def __str__(self):
        return f'"{self.titel}" av {self.forfattare}'

b = Bok("Mio, min Mio", "Astrid Lindgren")
print(b)   # anropar automatiskt __str__
```

### Exempel 9 — Klassattribut

Ett **klassattribut** delas av ALLA objekt av klassen (till skillnad från vanliga attribut, som är unika för varje objekt). Ett vanligt användningsområde är att räkna hur många objekt som skapats.

```python
class Bil:
    antal_bilar = 0   # klassattribut — delas av alla objekt

    def __init__(self, marke):
        self.marke = marke
        Bil.antal_bilar += 1

b1 = Bil("Volvo")
b2 = Bil("Saab")
b3 = Bil("Tesla")

print("Antal bilar skapade:", Bil.antal_bilar)
```

### Exempel 10 — Enkelt arv

Med **arv** kan en klass (subklass) återanvända attribut och metoder från en annan klass (superklass), och lägga till eller ändra egna. Här ärver `Katt` från `Djur`, men har sitt eget ljud.

```python
class Djur:
    def __init__(self, namn):
        self.namn = namn

    def lat(self):
        print(self.namn + " gör ett ljud.")

class Katt(Djur):          # Katt ärver från Djur
    def lat(self):          # denna metod skriver över Djurs version
        print(self.namn + " säger: Miau!")

d = Djur("Något djur")
k = Katt("Molly")

d.lat()
k.lat()
```

### Övning 10 — Enkelt arv: `Fordon` och `Bil`

Skapa en klass `Fordon` med attributet `namn` och en metod `beskrivning()` som skriver ut `"<namn> är ett fordon."`

Skapa sedan en klass `Bil` som **ärver** från `Fordon`, och som skriver över `beskrivning()` så den istället skriver ut `"<namn> är en bil."`

Skapa ett objekt av varje klass och anropa `beskrivning()` på båda.

```python
class Fordon:
    def __init__(self, namn):
        self.namn = namn

    def beskrivning(self):
        print(self.namn + " är ett fordon.")

class Bil(Fordon):
    def beskrivning(self):
        print(self.namn + " är en bil.")

f = Fordon("Något fordon")
b = Bil("Volvo")

f.beskrivning()
b.beskrivning()
```
